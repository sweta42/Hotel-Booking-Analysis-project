## 3. ***Data Wrangling***

### Data Wrangling Code

**Droping duplicates value**

````python
# 1.1 Check for missing values
miss_value[:10]
````

````python
# Remove duplicates
data.drop_duplicates(inplace=True)
````

````python
# After removing duplicates , count of total raws
data.shape
````

**replacing null values**

````python
# droping the company column as it has highest null values
data.drop(['company'], axis=1, inplace=True)
````

````python
# replacing null value in children column with 0 as there are only 4  missing values , assuming that family had 0 children
data['children'].fillna(0, inplace=True)
# replacing null value in agent column with 0 assuming those rooms were without company/agent
data['agent'].fillna(0, inplace=True)
# data[['children']] = data['children'].fillna(0)
````

````python
# replacing null values in country column as 'others' assuming that user didn't found their country name
data['country'].fillna('others', inplace = True)
````

````python
# checking that now there are no null values
data.isna().sum().sort_values(ascending=False)
````

````python
# there are some rows with total number of adults, children or babies eqal to zero
# This means there is no any booking were made
# so we can remove such rows

data.drop(data[data['adults']+data['babies']+data['children'] == 0].index, inplace = True)
````

````python
# Number or rows having with 0 in adult babies and children coloumn
zero_entry = data[data['adults']+data['babies']+data['children'] == 0].shape[0]
print(f"there are {zero_entry} rows which has zero value")
````

### changing data type

````python
# showing the info of data to check datatype
data.info()
````

````python
# We have seen that childern & agent column as datatype as float whereas it contains only int value, so change datatype as int64
data[['children', 'agent']] = data[['children', 'agent']].astype('int64')
````

````python
# for our understanding, in column 'is_canceled' we will replace the value from (0,1) to not_canceled, is canceled.
data['is_canceled'] = data['is_canceled'].replace([0,1], ['not canceled', 'is canceled'])
````

````python
# Same for 'is repeated guest' column
data['is_repeated_guest'] = data['is_repeated_guest'].replace([0,1], ['not repeated', 'repeated'])
````

## addition of new coloumn

````python
# We have created a col for total stays in nights by adding week night & weekend nights stay col.
# total stay in nights
data['total_stay_in_nights'] = data ['stays_in_week_nights'] + data ['stays_in_weekend_nights']
````

````python
# We have created a col for revenue using total stay* adr

data['revenue'] = data['total_stay_in_nights'] *data['adr']
````

````python
# Also, for information, we will add a column with total number of guests coming for each booking
data['total_guest'] = data['adults'] + data['children'] + data[ 'babies']
````

#### Chart - 10

### what is ADR accross different months?

````python
# Chart - 10 visualization code
# Using groupby funtion
bookings_by_months_df = data.groupby(['arrival_date_month', 'hotel']) ['adr'].mean().reset_index()

# Create month list
months = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']

# It will take the order of the month list in the dataframe along with values
bookings_by_months_df['arrival_date_month'] = pd.Categorical(bookings_by_months_df['arrival_date_month'], categories = months, ordered = True)

# Sorting values
bookings_by_months_df = bookings_by_months_df.sort_values('arrival_date_month')
````

````python
# Setting the chart size
plt.figure(figsize=(15,5))

# Plotting the values in a line chart
sns.lineplot(x=bookings_by_months_df['arrival_date_month'], y=bookings_by_months_df ['adr'], hue=bookings_by_months_df ['hotel'])

# Setting the labels and title
plt.title('ADR across each month', fontsize=20)
plt.xlabel('Month Name', fontsize=12)
plt.ylabel('ADR', fontsize=12)

#Show chart
plt.show()
````

I have used line plot in this case because it effectively shows the trend of ADR over time
